```python
def split_document_if_too_long(text, relations_df):
    max_length = 700 # 700 characters
    def sentence_spans(txt):
        tokens=nltk.sent_tokenize(txt)
        offset = 0
        for token in tokens:
            offset = txt.find(token, offset)
            yield token, offset, offset+len(token)
            offset += len(token)
    sentences = sentence_spans(text)

    documents = []
    start_offset = 0
    end_offset = 0
    for sentence in sentences:
        new_document_length = sentence[2] - start_offset
        if end_offset - start_offset > 0 and new_document_length > max_length:
            doc_text = text[start_offset: end_offset]
            doc_relations = []
            for row in relations_df.iloc:
                if row.event1_start >= start_offset and row.event1_end <= end_offset and row.event2_start >= start_offset and row.event2_end <= end_offset:
                    doc_relations.append({"event1_text": row["event1_text"], "event2_text": row["event2_text"],
                                         "event1_start": row["event1_start"] - start_offset, "event2_start": row["event2_start"] - start_offset,
                                         "event1_end": row["event1_end"] - start_offset, "event2_end": row["event2_end"] - start_offset,
                                         "class": row["class"]})
            documents.append({"text": doc_text, "relations": doc_relations})
            start_offset = end_offset
        end_offset = sentence[2]
    if end_offset - start_offset > 0:
        doc_text = text[start_offset: end_offset]
        doc_relations = []
        for row in relations_df.iloc:
            if row.event1_start >= start_offset and row.event1_end <= end_offset and row.event2_start >= start_offset and row.event2_end <= end_offset:
                doc_relations.append({"event1_text": row["event1_text"], "event2_text": row["event2_text"],
                                     "event1_start": row["event1_start"] - start_offset, "event2_start": row["event2_start"] - start_offset,
                                     "event1_end": row["event1_end"] - start_offset, "event2_end": row["event2_end"] - start_offset,
                                     "class": row["class"]})
        documents.append({"text": doc_text, "relations": doc_relations})
    return documents
```

```python
import json
from nltk.tokenize import word_tokenize
def convert_df(df, save_name):
    document_ids = set(df["document_id"])
    def spans(txt):
        tokens=nltk.word_tokenize(txt)
        offset = 0
        for token in tokens:
            offset = txt.find(token, offset)
            yield token, offset, offset+len(token)
            offset += len(token)
    
    def find_token(token_spans, index):
        for i, t in enumerate(token_spans):
            if t[1] <= index and t[2] >= index:
                return i, t
        return -1, None
    
    result = []
    for document_id in document_ids:
        all_relations = df[df["document_id"] == document_id]
        all_text = all_relations.iloc[0]["text"]
        split_documents = split_document_if_too_long(all_text, all_relations)
        for doc_idx, doc in enumerate(split_documents):
            relations = doc["relations"]
            text = doc["text"]
            # print(text)
            events = set([(r["event1_text"], r["event1_start"], r["event1_end"]) for r in relations] + [(r["event2_text"], r["event2_start"], r["event2_end"]) for r in relations])
            events_map = {}
            text_ind = 1
            event_ind = 1
            
            res_obj = {"tokens": word_tokenize(text)}
            token_spans = list(spans(text))
        
            events_tokens = []
            events_tokens_in_token_span_format = []
            for ind, event in enumerate(events):
                text, start, end = event
                start_token_idx, start_token = find_token(token_spans, start)
                end_token_idx, end_token = find_token(token_spans, end)
                events_tokens.append({"type": "event", "start": start_token_idx, "end": end_token_idx + 1})
                events_tokens_in_token_span_format.append((text, start, end))
            res_obj["entities"] = events_tokens
        
            relations_format = []
            for row in relations:
                head, _ = find_token(events_tokens_in_token_span_format, row["event1_start"])
                tail, _ = find_token(events_tokens_in_token_span_format, row["event2_start"])
                relations_format.append({"head": head, "tail": tail, "type": row["class"]})
            res_obj["relations"] = relations_format
            res_obj["orig_id"] = document_id
            result.append(res_obj)
    
    # Serializing json
    json_object = json.dumps(result, indent=0)
     
    # Writing to sample.json
    with open(save_name, "w") as outfile:
        outfile.write(json_object)
```

```python
df = read_i2b2(full_text=True, use_test_files=False)
train_df, val_df, test_df = split_data(df, oversample=False)
convert_df(train_df, "i2b2_all_train.json")
convert_df(val_df, "i2b2_all_val.json")
df_test = read_i2b2(full_text=True, use_test_files=True)
convert_df(df_test, "i2b2_all_test.json")
```
